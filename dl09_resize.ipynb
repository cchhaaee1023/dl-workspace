```python
import platform
import matplotlib.pyplot as plt

import torch
import torchvision.transforms.v2 as transforms
import torchvision.io as io
```

```python
# OS에 따른 한글 폰트 설정
if platform.system() == 'Windows':
    plt.rc('font', family='Malgun Gothic')  # 윈도우: 맑은 고딕
elif platform.system() == 'Darwin':
    plt.rc('font', family='AppleGothic')    # 맥: 애플 고딕
else:
    plt.rc('font', family='NanumBarunGothic') # 리눅스

# 마이너스 기호 깨짐 방지
plt.rc('axes', unicode_minus=False)
```

```python
# 이미지 경로
image_path = "data/four.jpg"  # 이미지 경로

# 1. PyTorch를 사용해 이미지를 텐서 형태로 직접 로드 (C, H, W / uint8)
original_tensor = io.read_image(image_path)
print(f'original_tensor.shape: {original_tensor.shape}')
c, h, w = original_tensor.shape
print(f"Original shape (C, H, W): {original_tensor.shape}")

# 2. PyTorch Transform을 사용한 리사이즈 (224x224)
resize_transform = transforms.Resize((224, 224), antialias=True)
resized_tensor = resize_transform(original_tensor)
print(f'resized_tensor: {resized_tensor.shape}')

# 3. Matplotlib 시각화를 위해 (C, H, W) -> (H, W, C)로 넘파이/텐서 차원 변경 (.permute 사용)
original_display = original_tensor.permute(1, 2, 0)
resized_display = resized_tensor.permute(1, 2, 0)

# 4. 시각화
plt.figure(figsize=(10, 5))

# 원본 이미지
plt.subplot(1, 2, 1)
plt.imshow(original_display, cmap='gray')
plt.title(f"Original(원본) Image ({w}x{h})")
plt.axis("off")

# 리사이즈 이미지
plt.subplot(1, 2, 2)
plt.imshow(resized_display, cmap='gray')
plt.title("Resized Image (224x224)")
plt.axis("off")

plt.tight_layout()
plt.show()
```
