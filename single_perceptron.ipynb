```python
def single_layer_perceptron(x1, x2, w1, w2, b):
    """
    단층 퍼셉트론 연산 함수
    x1, x2 : 입력값
    w1, w2 : 가중치 (Weight)
    b      : 편향 (Bias)
    """
    # 1. 입력값과 가중치의 곱에 편향을 더함 (순입력 총합)
    total_sum = (x1 * w1) + (x2 * w2) + b
    
    # 2. 활성화 함수 (임계값 0을 기준으로 한 계단 함수 적용)
    if total_sum > 0:
        return 1
    else:
        return 0

# 진리표 테스트를 위한 4가지 입력 조합
inputs = [(0, 0), (1, 0), (0, 1), (1, 1)]

# --- [1] AND 게이트 테스트 ---
print("--- [AND 게이트 진리표] ---")
w1, w2, b = 0.5, 0.5, -0.7
for x1, x2 in inputs:
    output = single_layer_perceptron(x1, x2, w1, w2, b)
    print(f"입력: ({x1}, {x2}) -> 출력: {output}")

print("\n--- [OR 게이트 진리표] ---")
# --- [2] OR 게이트 테스트 ---
w1, w2, b = 0.5, 0.5, -0.2
for x1, x2 in inputs:
    output = single_layer_perceptron(x1, x2, w1, w2, b)
    print(f"입력: ({x1}, {x2}) -> 출력: {output}")

print("\n--- [NAND 게이트 진리표] ---")
# --- [3] NAND 게이트 테스트 ---
w1, w2, b = -0.5, -0.5, 0.7
for x1, x2 in inputs:
    output = single_layer_perceptron(x1, x2, w1, w2, b)
    print(f"입력: ({x1}, {x2}) -> 출력: {output}")
```
