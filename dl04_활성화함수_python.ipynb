```python
import numpy as np

# 1. 활성화 함수 및 미분 정의
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    # sigmoid(x)의 미분값: sigmoid(x) * (1 - sigmoid(x))
    # 이미 활성화 함수를 거친 출력값(a)이 들어온다고 가정하면: a * (1 - a)
    return x * (1 - x)

# 2. 신경망 클래스 정의
class NeuralNetwork:
    def __init__(self, input_size, hidden_size, output_size, learning_rate=0.5):
        self.learning_rate = learning_rate

        # 가중치 및 편향 난수 초기화 (가중치 모양: [입력, 출력])
        np.random.seed(42)  # 재현성을 위한 시드 고정
        self.W1 = np.random.randn(input_size, hidden_size)
        self.b1 = np.zeros((1, hidden_size))

        self.W2 = np.random.randn(hidden_size, output_size)
        self.b2 = np.zeros((1, output_size))

    def forward(self, X):
        """순전파 (Forward Propagation)"""
        # 은닉층 연산: z1 = X*W1 + b1, a1 = sigmoid(z1)
        self.z1 = np.dot(X, self.W1) + self.b1
        self.a1 = sigmoid(self.z1)

        # 출력층 연산: z2 = a1*W2 + b2, a2 = sigmoid(z2)
        self.z2 = np.dot(self.a1, self.W2) + self.b2
        self.a2 = sigmoid(self.z2)

        return self.a2

    def backward(self, X, y):
        """역전파 (Backpropagation) & 경사하강법"""
        m = X.shape[0]  # 샘플 개수

        # [출력층 오차 및 기울기 계산]
        # 1) 손실함수(MSE) 미분: dL/da2 = (a2 - y)
        loss_grad = (self.a2 - y)
        # 2) 출력층 오차 신호: delta2 = dL/dz2 = (a2 - y) * sigmoid'(z2)
        delta2 = loss_grad * sigmoid_derivative(self.a2)

        # 3) W2, b2에 대한 기울기(Gradient) 계산
        dW2 = np.dot(self.a1.T, delta2) / m
        db2 = np.sum(delta2, axis=0, keepdims=True) / m

        # [은닉층 오차 및 기울기 계산 (연쇄 법칙)]
        # 4) 은닉층으로 역전파되는 오차: delta1 = (delta2 * W2^T) * sigmoid'(z1)
        delta1 = np.dot(delta2, self.W2.T) * sigmoid_derivative(self.a1)

        # 5) W1, b1에 대한 기울기(Gradient) 계산
        dW1 = np.dot(X.T, delta1) / m
        db1 = np.sum(delta1, axis=0, keepdims=True) / m

        # [가중치 및 편향 업데이트 (경사하강법)]
        self.W2 -= self.learning_rate * dW2
        self.b2 -= self.learning_rate * db2
        self.W1 -= self.learning_rate * dW1
        self.b1 -= self.learning_rate * db1

    def train(self, X, y, epochs=10000):
        """학습 루프"""
        for epoch in range(epochs):
            # 순전파
            output = self.forward(X)

            # 역전파 및 가중치 업데이트
            self.backward(X, y)

            # 1000회마다 손실값(MSE) 출력
            if (epoch + 1) % 1000 == 0:
                loss = np.mean((output - y) ** 2)
                print(f"Epoch {epoch + 1:5d} | Loss: {loss:.6f}")

# -------------------------------------------------------------
# 3. 데이터셋 준비 (XOR 문제)
# Inputs: [0,0], [0,1], [1,0], [1,1]
# Targets:   0,      1,      1,      0
X = np.array([[0, 0],
                [0, 1],
                [1, 0],
                [1, 1]])

y = np.array([[0],
                [1],
                [1],
                [0]])

# 4. 모델 생성 및 학습 (입력 2개 -> 은닉층 4개 -> 출력 1개)
nn = NeuralNetwork(input_size=2, hidden_size=4, output_size=1, learning_rate=5.0)

print("=== 학습 시작 ===")
nn.train(X, y, epochs=10000)

print("\n=== 최종 예측 결과 ===")
predictions = nn.forward(X)
for i in range(len(X)):
    print(f"입력: {X[i]} -> 예측값: {predictions[i][0]:.4f} (실제값: {y[i][0]})")
```
